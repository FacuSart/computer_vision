## 3. Bajar el dataset

En Roboflow Universe: abrí el dataset → **Download Dataset** → formato **YOLOv11** →
"show download code" y pegá acá ese snippet (trae tu API key, workspace y proyecto).

```python
from getpass import getpass
from roboflow import Roboflow

rf = Roboflow(api_key=getpass("Roboflow API key: "))
project = rf.workspace("imaging-oltxn").project("detection-stpgg")
dataset = project.version(1).download("yolov11")

DATA_YAML = f"{dataset.location}/data.yaml"
print(DATA_YAML)
```

## 4. Entrenar

```python
!python src/train.py --data {DATA_YAML} --epochs 10 --name prueba
```
